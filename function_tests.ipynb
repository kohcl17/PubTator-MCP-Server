# Case for LLMs

- While we can manually look up possible autoimmune diseases to do a search on, we can easily ask an LLM for potential autoimmune disease search terms to look for

- The LLM can then pass these onto PubTator3 to look up its entity id

- Evidence: Looking up gene relation to autoimmune_diseases id is too broad and doesn't usually yield great results. Looking up only SLE limits the scope substantially.

```python
entity_1 = api.find_entity_id("OLR1", "gene", limit = 1)[0]['_id']
entity_2 = api.find_entity_id("systemic lupus", "disease", limit = 1)[0]['_id']
```

```python
result = api.search(f'{entity_1} AND {entity_2}', max_pages = 10)
result = list(str(r['results'][0]['pmid']) for r in result)
```

```python
pubs = api.export_publications(result, format = 'biocjson', full_text=False, extract_text=True)
```

```python
relations = api.search_relations(entity_1, entity2 = entity_2, max_pages = 20)
```

```python
relations_pmids = list(str(r['results'][0]['pmid']) for r in relations)
```

```python
api.export_publications(relations_pmids, extract_text=True)
```
